## Step 1: Define The Problem.

You're currently a data scientist working at a university. A professor of psychology is attempting to study the relationship between personalities and left-handedness. They have tasked you with gathering evidence so that they may publish.

As a data scientist, you know that any real data science problem must be **specific** and **conclusively answerable**. For example:
- Bad data science problem: "What is the link between obesity and blood pressure?"
    - This is vague and is not conclusively answerable. That is, two people might look at the conclusion and one may say "Sure, the problem has been answered!" and the other may say "The problem has not yet been answered."
- Good data science problem: "Does an association exist between obesity and blood pressure?"
    - This is more specific and is conclusively answerable. The problem specifically is asking for a "Yes" or "No" answer. Based on that, two independent people should both be able to say either "Yes, the problem has been answered" or "No, the problem has not yet been answered."
- Excellent data science problem: "As obesity increases, how does blood pressure change?"
    - This is very specific and is conclusively answerable. The problem specifically seeks to understand the effect of one variable on the other.

## Step 4 & 5 Modeling: $k$-nearest neighbors

### Train-test split your data:

Your explanatory variables should be all Q columns.

```python
from sklearn.model_selection import train_test_split

X = df[[f"Q{i}" for i in range(1, 45)]]
y = df["hand"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
```

### Create and fit four separate $k$-nearest neighbors models:

### Model 1: one with $k = 3$:

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
```

#### Model 2: one with $k = 5$

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
```

#### Model 3: one with $k = 15$

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=15)
knn.fit(X_train, y_train)
```

#### Model 4: one with $k = 25$:

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=25)
knn.fit(X_train, y_train)
```

### Evaluate your models:

Evaluate each of your four models on the training and testing sets, and
interpret the four scores.

Are any of your models overfit or underfit?

Do any of your models beat the baseline accuracy rate?

```python
from sklearn.neighbors import KNeighborsClassifier
import pandas as pd

results = []

for k in [5, 10, 15, 25]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)

    results.append({
        "k": k,
        "Train Accuracy": knn.score(X_train, y_train),
        "Test Accuracy": knn.score(X_test, y_test)
    })

pd.DataFrame(results)
```

```python
from sklearn.metrics import classification_report
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=25)
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)

print(classification_report(y_test, y_pred, zero_division=0))
```

```python
for k in [5, 10, 15, 25]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)

    print(f"\nKNN with k={k}")
    print(classification_report(y_test, y_pred, zero_division=0))
```

```python
#**Evaluation of KNN Models**

All four KNN models achieved approximately 85% test accuracy, which is similar to the baseline accuracy of 84.9%.

The model with k=5 performed slightly better at identifying left-handed individuals, with a recall of 4%. The other models (k=10, 15, and 25) failed to identify any left-handed individuals.

The dataset is highly imbalanced, with right-handed individuals representing the majority. Therefore, high accuracy does not necessarily indicate good classification performance.

**Conclusion:** None of the four KNN models is effective for predicting left-handedness. The k=5 model identifies a small number of left-handed individuals, but its recall is still very low.
```
